```python
%matplotlib widget
import os, re, json, pathlib
import warnings                 # ← чтобы тихо сообщать, когда probs нет
from typing import List, Dict, Any

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from ipywidgets import (
    Button,
    HBox,
    VBox,
    Dropdown,
    Output,
    FloatText,
)
from IPython.display import display, clear_output
#from scipy.signal import find_peaks


class SpectrumAnnotator:
    """Мини‑GUI для просмотра спектрограмм и разметки пиков без мышиного
    выделения прямоугольников.

    Видимые регионы отображаются красной рамкой + подписью. Разметку можно
    добавлять:
      * автоматически через *Auto Peaks*;
      * вручную по диапазону частот через поля *F start / F end*.

    Выбор прямоугольника мышью **отключён**, кнопка *Add Region* и
    `RectangleSelector` убраны.
    """

    # ---------------------------------------------------------------------
    #                               init
    # ---------------------------------------------------------------------

    def __init__(self, data_dir: str | pathlib.Path, noise_dir: str | pathlib.Path | None = None):
        self.data_dir = pathlib.Path(data_dir)
        self.noise_dir = pathlib.Path(noise_dir) if noise_dir else None

        self.files: List[str] = self._get_sorted_files()
        if not self.files:
            raise FileNotFoundError("No CSV files found in data_dir.")

        self.current_file_idx: int = 0
        self.freqs: np.ndarray | None = None
        self.times: np.ndarray | None = None
        self.data: np.ndarray | None = None

        # --- probability-кривая ---
        self.prob_freqs: np.ndarray | None = None
        self.prob_vals:  np.ndarray | None = None

        
        self.noise_freqs: np.ndarray | None = None
        self.noise_times: np.ndarray | None = None
        self.noise_data: np.ndarray | None = None
        self.regions: List[Dict[str, Any]] = []

        self.annotations_path = self.data_dir / "annotations.json"

        self.output = Output()
        self.fig: plt.Figure | None = None
        self.ax_signal = None
        self.ax_noise = None
        self.img_signal = None
        self.img_noise = None

        self._build_widgets()
        self._load_current_file()
        self._setup_ui()

    # ------------------------------------------------------------------
    #                       File helpers
    # ------------------------------------------------------------------

    def _get_sorted_files(self) -> List[str]:
        def _f_mhz(name: str) -> int:
            m = re.search(r"(\d+)MHz", name)
            return int(m.group(1)) if m else 0

        return sorted(
            [f for f in os.listdir(self.data_dir) if f.lower().endswith(".csv")],
            key=_f_mhz,
        )

    def _load_current_file(self):
        path = self.data_dir / self.files[self.current_file_idx]
        df = pd.read_csv(path)
        self.times = df.iloc[:, 0].to_numpy(float)
        self.freqs = df.columns[1:].to_numpy(float)
        self.data = df.iloc[:, 1:].to_numpy(float)
        self._load_regions()

        if self.noise_dir is not None:
            nfile = self.noise_dir / self.files[self.current_file_idx]
            if nfile.exists():
                ndf = pd.read_csv(nfile)
                self.noise_times = ndf.iloc[:, 0].to_numpy(float)
                self.noise_freqs = ndf.columns[1:].to_numpy(float)
                self.noise_data = ndf.iloc[:, 1:].to_numpy(float)
            else:
                self.noise_times = self.noise_freqs = self.noise_data = None
        else:
            self.noise_times = self.noise_freqs = self.noise_data = None

                # --- probs ---
        pfile = self.data_dir / "probs" / f"{path.stem}_probs.csv"
        if pfile.exists():
            pdf = pd.read_csv(pfile)
            self.prob_freqs = pdf["freq_MHz"].to_numpy(float)
            self.prob_vals  = pdf["p_avg"].to_numpy(float)
        else:
            self.prob_freqs = self.prob_vals = None
            warnings.warn(f"probability file not found: {pfile.name}", stacklevel=2)

    def _load_regions(self):
        self.regions = []
        if self.annotations_path.exists():
            with open(self.annotations_path, "r", encoding="utf-8") as f:
                ann = json.load(f)
            self.regions = ann.get(self.files[self.current_file_idx], [])

    # ------------------------------------------------------------------
    #                           Widgets
    # ------------------------------------------------------------------

    def _build_widgets(self):
        # выбор файла
        self.dd_files = Dropdown(options=self.files, description="File:", layout={"width": "450px"})
        self.dd_files.observe(self._on_file_change, names="value")

        # навигация
        self.btn_prev = Button(description="◀")
        self.btn_next = Button(description="▶")
        self.btn_prev.on_click(self._prev_file)
        self.btn_next.on_click(self._next_file)

        # тип региона (используется авто/ручной ввод)
        self.dd_type = Dropdown(options=["SIGNAL", "SIGNAL_NORMAL", "SIGNAL_GOOD", "SIGNAL_MAIN", "NOISE_STRONG"], description="Type:")

        # удаление / сброс
        self.btn_del = Button(description="Delete Last")
        self.btn_reset = Button(description="Reset")
        self.btn_del.on_click(self._del_last)
        self.btn_reset.on_click(self._reset_regions)

        # ручной диапазон частот
        self.ft_start = FloatText(description="F start, MHz:", layout={"width": "160px"})
        self.ft_end = FloatText(description="F end, MHz:", layout={"width": "160px"})
        self.btn_add_freq = Button(description="Add by freq")
        self.btn_add_freq.on_click(self._add_freq_region)

        # авто‑пики
        self.ft_visible_frac = FloatText(value=0.3, min=0, max=1, step=0.05,
                                         description="visible_fraction:", layout={"width": "220px"})
        #self.btn_auto = Button(description="Auto Peaks")
        #self.btn_auto.on_click(self._auto_peaks)

        # сохранение
        self.btn_save = Button(description="Save", button_style="success")
        self.btn_save.on_click(self._save_regions)

    def _setup_ui(self):
        controls = VBox(
            [
                self.dd_files,
                HBox([self.btn_prev, self.btn_next]),
                HBox([self.dd_type, self.btn_del, self.btn_reset]),
                HBox([self.ft_start, self.ft_end, self.btn_add_freq]),
                HBox([self.ft_visible_frac]),
                self.btn_save,
            ]
        )
        display(controls, self.output)
        self._init_figure()
        self._update_plot()

    # ------------------------------------------------------------------
    #                       Matplotlib part
    # ------------------------------------------------------------------

    def _init_figure(self):
        # определяем, нужны ли noise/probs панели
        show_noise = self.noise_dir is not None and self.noise_data is not None
        show_probs = self.prob_vals is not None
    
        n_rows = 1 + int(show_noise) + int(show_probs)
        heights = [1] * n_rows
    
        with self.output:
            clear_output(wait=True)
            self.fig, axes = plt.subplots(
                n_rows, 1,
                figsize=(13, 5 * n_rows),
                sharex=True,
                gridspec_kw={"height_ratios": heights},
            )
    
            # если axes не итерируем (n_rows == 1)
            if n_rows == 1:
                axes = [axes]
    
            # --- сигнал ---
            self.ax_signal = axes[0]
            self.ax_signal.set_title("Signal")
            self.ax_signal.set_ylabel("Time (s)")
            self.ax_signal.set_xlabel("Frequency (MHz)")
            self.img_signal = self.ax_signal.imshow(
                np.zeros((2, 2)), aspect="auto", origin="upper"
            )
    
            row = 1
    
            # --- noise ---
            if show_noise:
                self.ax_noise = axes[row]
                self.ax_noise.set_title("Reference: Noise only")
                self.ax_noise.set_ylabel("Time (s)")
                self.ax_noise.set_xlabel("Frequency (MHz)")
                self.img_noise = self.ax_noise.imshow(
                    np.zeros((2, 2)), aspect="auto", origin="upper"
                )
                row += 1
            else:
                self.ax_noise = self.img_noise = None
    
            # --- probs ---
            if show_probs:
                self.ax_probs = axes[row]
                self.ax_probs.set_title("Average probability")
                self.ax_probs.set_ylabel("p_avg")
                self.ax_probs.set_xlabel("Frequency (MHz)")
                # линия-заглушка; данные обновим в _update_plot
                (self.line_probs,) = self.ax_probs.plot([], [], linewidth=1)
                self.ax_probs.grid(True)
            else:
                self.ax_probs = self.line_probs = None
    
            plt.tight_layout()
            plt.show()

    # ------------------------------------------------------------------
    #                     Plot helpers
    # ------------------------------------------------------------------

    def _draw_region(self, ax, r: Dict[str, Any]):
        """Красная рамка + подпись."""
        rect = plt.Rectangle(
            (r["freq_start"], r["time_start"]),
            r["freq_end"] - r["freq_start"],
            r["time_end"] - r["time_start"],
            fill=False,
            edgecolor="red" if r.get("type") != "NOISE_STRONG" else "blue",
            linewidth=1.5,
        )
        ax.add_patch(rect)
        ax.text(
            r["freq_start"],
            (r["time_start"] + r["time_end"]) / 2,
            r.get("type", "SIGNAL"),
            color="red" if r.get("type") != "NOISE_STRONG" else "blue",
            fontsize=8,
            va="center",
            bbox=dict(facecolor="white", alpha=0.7, pad=2, edgecolor="none"),
        )

    def _update_plot(self):
        # цветовая шкала
        vmin, vmax = (
            np.percentile(np.concatenate([self.data.ravel(), self.noise_data.ravel()]), [5, 95])
            if self.noise_data is not None
            else np.percentile(self.data, [5, 95])
        )

        self.img_signal.set_data(self.data)
        self.img_signal.set_extent([self.freqs[0], self.freqs[-1], self.times[0], self.times[-1]])
        self.img_signal.set_clim(vmin, vmax)

        if self.noise_data is not None:
            self.img_noise.set_data(self.noise_data)
            self.img_noise.set_extent([self.noise_freqs[0], self.noise_freqs[-1], self.noise_times[0], self.noise_times[-1]])
            self.img_noise.set_clim(vmin, vmax)

        # очищаем старые патчи/тексты
        for artist in list(self.ax_signal.patches) + list(self.ax_signal.texts):
            artist.remove()

        for r in self.regions:
            self._draw_region(self.ax_signal, r)

                # --- probability curve ---
        if self.line_probs is not None and self.prob_vals is not None:
            self.line_probs.set_data(self.prob_freqs, self.prob_vals)
            # обновляем лимиты
            self.ax_probs.set_xlim(self.prob_freqs[0], self.prob_freqs[-1])
            self.ax_probs.set_ylim(0, self.prob_vals.max() * 1.05)


        self.fig.canvas.draw_idle()

    # ------------------------------------------------------------------
    #                       Region editing
    # ------------------------------------------------------------------

    def _save_regions(self, _):
        ann = {}
        if self.annotations_path.exists():
            with open(self.annotations_path, "r", encoding="utf-8") as f:
                ann = json.load(f)
        ann[self.files[self.current_file_idx]] = self.regions
        with open(self.annotations_path, "w", encoding="utf-8") as f:
            json.dump(ann, f, indent=2, ensure_ascii=False)
        print(f"💾 Saved {len(self.regions)} region(s) → {self.annotations_path.name}")

    def _del_last(self, _):
        if self.regions:
            self.regions.pop()
            self._update_plot()

    def _reset_regions(self, _):
        self.regions.clear()
        self._update_plot()

    # --------------------------------------------------------------
    #                 Manual frequency range add
    # --------------------------------------------------------------

    def _add_freq_region(self, _):
        f1, f2 = sorted([self.ft_start.value, self.ft_end.value])
        if f1 == f2:
            print("⚠️ Zero-width range skipped.")
            return
        region = {
            "freq_start": f1,
            "freq_end": f2,
            "time_start": float(self.times[0]),
            "time_end": float(self.times[-1]),
            "type": self.dd_type.value,
        }
        self.regions.append(region)
        self._update_plot()

    # --------------------------------------------------------------
    #                 Automatic peak detection
    # --------------------------------------------------------------
    # ---------------------- полу‑автомат ----------------------
    """def _auto_peaks(self, _):
        if self.data is None:
            return
        mean_spectrum = self.data.mean(axis=0)
        # элементарная выборка порога: 5 % от максимума
        prominence = mean_spectrum.max() * 0.05
        peaks, _ = find_peaks(mean_spectrum, prominence=prominence)
        if not len(peaks):
            print("❗️ Peeks not found – попробуйте снизить prominence.")
            return

        max_amp = mean_spectrum[peaks].max()
        visible_thr = self.ft_visible_frac.value * max_amp

        # ширина области = один биновый шаг
        df = self.freqs[1] - self.freqs[0] if len(self.freqs) > 1 else 0.1
        added = 0
        for p in peaks:
            f = self.freqs[p]
            amp = mean_spectrum[p]
            region_type = "SIGNAL_MAIN" if amp >= visible_thr else "SIGNAL"
            region = {
                "freq_start": f - df / 2,
                "freq_end": f + df / 2,
                "time_start": float(self.times[0]),
                "time_end": float(self.times[-1]),
                "type": region_type,
            }
            self.regions.append(region)
            added += 1

        print(f"🔍 Added {added} auto‑detected peaks (visible_fraction={self.ft_visible_frac.value})")
        self._update_plot()
        """

    # ---------------------- ручной ввод частот ----------------------
    def _add_freq_region(self, _):
        f1, f2 = sorted([self.ft_start.value, self.ft_end.value])
        if f1 == f2:
            print("⚠️  Диапазон нулевой ширины – не добавлен.")
            return
        region = {
            "freq_start": f1,
            "freq_end": f2,
            "time_start": float(self.times[0]),
            "time_end": float(self.times[-1]),
            "type": self.dd_type.value,
        }
        self.regions.append(region)
        self._update_plot()

    # ---------------------- навигация по файлам ----------------------
    def _on_file_change(self, change):
        if change["name"] == "value":
            self.current_file_idx = self.files.index(change["new"])
            self._load_current_file()
            self._update_plot()

    def _prev_file(self, _):
        if self.current_file_idx > 0:
            self.current_file_idx -= 1
            self.dd_files.value = self.files[self.current_file_idx]

    def _next_file(self, _):
        if self.current_file_idx < len(self.files) - 1:
            self.current_file_idx += 1
            self.dd_files.value = self.files[self.current_file_idx]
```
